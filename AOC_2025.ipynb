# Day 1: Secret Entrance

Link: https://adventofcode.com/2025/day/1

```python
rotations = open('/content/drive/MyDrive/aoc_2025/day1_input_25.txt').read().splitlines()

password = 0
pointer = 50

for move in rotations:
  direction = move[0]
  amount = int(move[1:])

  if direction == 'L':
    pointer = (pointer - amount) % 100
  else:
    pointer = (pointer + amount) % 100

  if pointer == 0:
    password += 1

print(password)
```

--- Part Two ---

```python
rotations = open('/content/drive/MyDrive/aoc_2025/day1_input_25.txt').read().splitlines()

password = 0
pointer = 50

for move in rotations:
  direction = move[0]
  amount = int(move[1:])

  if direction == 'R':
    step = 1
  else:
    step = -1

  first_hit = (-pointer * step) % 100

  if first_hit == 0:
    first_hit = 100

  if first_hit <= amount:
    password += 1 + (amount - first_hit) // 100

  pointer = (pointer + step * amount) % 100

print(password)
```

# Day 2: Gift Shop

Link: https://adventofcode.com/2025/day/2

```python
def split_number_by_digits(number):
  """
  Splits a number into two integers based on its digits.
  If the number has an odd number of digits, the first half
  will contain the extra digit.
  """
  s_num = str(number)
  midpoint = len(s_num) // 2

  first_half_str = s_num[:midpoint]
  second_half_str = s_num[midpoint:]

  # Convert back to integers
  first_half = int(first_half_str)
  second_half = int(second_half_str)

  return first_half, second_half

ranges = open('/content/drive/MyDrive/aoc_2025/day2_input_25.txt').read().split(',')

invalid_ids = 0

for r in ranges:
  start, stop = r.split('-')
  start = int(start)
  stop = int(stop)

  for n in __builtins__.range(start, stop + 1):
    # Invalid IDs must have an even number of digits
    if len(str(n)) % 2 == 0:
      x, y = split_number_by_digits(n)

      if x == y:
        invalid_ids += n

print(invalid_ids)
```

--- Part Two ---

```python
import re

def has_repeated_regex(number):
    """
    Checks if the entire number string is formed by a sequence of digits
    repeated at least twice.

    Args:
        number (int or str): The number to check.

    Returns:
        bool: True if the number is an invalid ID, False otherwise.
    """
    s = str(number)
    # Use re.fullmatch to ensure the entire string matches the pattern.
    # (.+?) - non-greedy capture of one or more characters (the repeating sequence)
    # \1+   - the captured group repeated one or more times (at least twice in total)
    return bool(re.fullmatch(r"(.+?)\1+", s))

ranges = open('/content/drive/MyDrive/aoc_2025/day2_input_25.txt').read().split(',')

invalid_ids = 0

for r in ranges:
  start, stop = r.split('-')
  start = int(start)
  stop = int(stop)

  for n in __builtins__.range(start, stop + 1):
    if has_repeated_regex(n): # Use the updated regex check
      invalid_ids += n

print(invalid_ids)
```

# Day 3: Lobby
Link: https://adventofcode.com/2025/day/3


```python
batteries = open('/content/drive/MyDrive/aoc_2025/day3_input_25.txt').read().splitlines()

total_joltage = 0

for b in batteries:
  max_joltage = 0

  for i in range(len(b)):
    for j in range(i + 1, len(b)):
      current_joltage_str = b[i] + b[j]
      current_joltage = int(current_joltage_str)
      if current_joltage > max_joltage:
        max_joltage = current_joltage

  total_joltage += max_joltage

print(total_joltage)
```

```python
batteries = open('/content/drive/MyDrive/aoc_2025/day3_input_25.txt').read().splitlines()

total_joltage = 0

def find_largest_number_by_removing_digits(s, k):
    n = len(s)
    if k == 0:
        return "0"
    if k > n:
        # This case should not happen based on problem constraints
        # but defensively return the original string if we need more digits than available.
        return s

    digits_to_remove = n - k
    result = [] # This will act as a monotonic stack to build the largest number

    for digit in s:
        # While the result is not empty, and the last digit in result is less than the current digit,
        # AND we still have digits to remove, pop from result.
        # This greedy step removes smaller digits to make space for larger ones earlier.
        while result and result[-1] < digit and digits_to_remove > 0:
            result.pop()
            digits_to_remove -= 1
        result.append(digit)

    # If we still have removals left, it means the remaining digits in 'result' are in decreasing
    # order (or all equal), so we remove the smallest ones from the end.
    while digits_to_remove > 0:
        result.pop()
        digits_to_remove -= 1

    # Ensure the final result has exactly k digits. This is defensive;
    # the algorithm typically yields k digits if n >= k.
    return "".join(result[:k])

for b in batteries:
  largest_12_digit_str = find_largest_number_by_removing_digits(b, 12)
  max_current_bank_joltage = int(largest_12_digit_str)
  total_joltage += max_current_bank_joltage

print(total_joltage)
```
